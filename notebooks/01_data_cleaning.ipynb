```python
import pandas as pd
import numpy as np
from pathlib import Path

# Initialize root directory
PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / 'data' / '01_raw' / 'adult.csv'

# Import data
df = pd.read_csv(DATA_PATH)
```

```python
print("Original dataset shape:", df.shape)
display(df.head())
```

```python
for col in df.select_dtypes(include="object").columns: # remove whitespace around categorical values
    df[col] = df[col].str.strip()

df = df.replace("?", np.nan) # replace ? with actual missing values

print(df.isnull().sum().sort_values(ascending=False)) # inspect missing values
```

```python
df_clean = df.dropna().copy()
df_clean = df_clean.drop_duplicates().copy()
print("Rows before cleaning:", len(df))
print("Rows after removing missing values and duplicates:", len(df_clean))
```

```python
# standardizing the income labels 
df_clean["income"] = (
    df_clean["income"]
    .astype(str)
    .str.strip()
    .str.rstrip(".")
)

print(df_clean["income"].value_counts())
```

```python
from pathlib import Path

output_dir = Path("../data") # save data folder in project
output_dir.mkdir(parents=True, exist_ok=True)

df_clean.to_csv(output_dir / "adult_cleaned.csv", index=False)

print("Final shape:", df_clean.shape)
```
